```python
from sklearn.model_selection import train_test_split
#syntax train_test_split(X, y, test_size=..., train_size=..., random_state=..., shuffle=...)
X_train, X_test, y_train, y_test = train_test_split(
    df.drop('Purchased', axis=1), #removing target column 
    df['Purchased'], 
    test_size=0.3, 
    random_state=42
    
)
```

```python
from sklearn.preprocessing import StandardScaler
scalar=StandardScaler()
#fit the scalar to the train set, it will learn the parameters 
scalar.fit(X_train) #mean std haru yeta store hnxa 
#transform train and test splits 
X_train_scaled=scalar.transform(X_train)
X_test_scaled=scalar.transform(X_test)
```
