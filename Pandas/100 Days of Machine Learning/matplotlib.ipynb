### MatplotLib Full Course One Shot

```python
%matplotlib widget
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import style
style.use("ggplot")
```

```python
x_data=np.random.random(50)*100
y_data=np.random.random(50)*100
```

```python
#scatter Plot 
plt.figure()
plt.scatter(x_data,y_data, c='red',marker="*")
plt.show()
```

```python
years=[2006 + x for x in range(16)]
weights=[50,56,51,52,49,56,54,48,49,51,56,57,48,50,51,54]
plt.figure()
plt.plot(years,weights,c='green',lw='1',linestyle="--")
```

```python
#Bar Plots 
x =['C++','C#','Python',"Rust","Java"]
votes=[20,50,140,45,15]
plt.figure()
plt.bar(x,votes, color="red", align="edge")
```

```python
#histogram 
ages=np.random.normal(20,1.5,1000)
plt.figure()
plt.hist(ages,bins=[ages.min(), 18 ,21, ages.max()],cumulative=True)
```

```python
plt.figure()
plt.pie(votes,labels=x,autopct="%.2f%%",startangle=90)
plt.show()
```

```python

heights=np.random.normal(172,8,300)
plt.figure()
plt.boxplot(heights)
plt.show()
```

###Customization Of The Plots

```python
income=np.random.randint(50,70,16)
income_ticks=list(range(48,72,2))
plt.figure()
plt.plot(years,income)
plt.title("Average Income of People Of Nepal",fontsize=12,fontname="Poppins")
plt.xlabel("Years in AD")
plt.ylabel('Income in Thousands')
plt.yticks(income_ticks, [f"Npr{x}k" for x in income_ticks])
```

plt.yticks(positions, labels)
positions → where ticks appear (numbers on axis)
labels → what text is shown at those positions

equivalent code 
labels = []
for x in income_ticks:
    labels.append(f"Npr{x}k")


### Legends And Multiplots 

```python
tesla=list(np.random.randint(100,105,8))
nasa=list(np.random.randint(100,105,8))
spacex=list(np.random.randint(100,105,8))
plt.figure()
plt.plot(tesla,label="tesla")
plt.plot(nasa,label='nasa')
plt.plot(spacex,label="spacex")
plt.legend(loc="best")
```

```python
plt.figure()
plt.pie(votes,labels=None)
plt.legend(labels=x,loc="best")
```

```python
plt.figure(1)
plt.plot(tesla,label="tesla")
plt.legend(loc="best")
plt.figure(2)
plt.plot(nasa,label='nasa')
plt.legend(loc="best")
plt.figure(3)
plt.plot(spacex,label="spacex")
plt.legend(loc="best")
plt.show()
```

```python
#Subplots 

x1=np.arange(20)
plt.figure()
fig ,axis = plt.subplots(2,2)
axis[0,0].plot(x1,np.sin(x1))
axis[0,0].set_title("Sine Plot")

axis[1,1].plot(x1,np.cos(x1))
axis[1,1].set_title("cosine Plot")
plt.tight_layout()

```

### ADVANCED 3D Plotting using Matplotlib

```python
plt.figure()
ax = plt.axes(projection="3d")

x = np.random.random(100)
y = np.random.random(100)
z = np.random.random(100)

ax.scatter(x, y, z)
ax.set_title("3D Plot")
ax.set_xlabel("x axis")
plt.show()
```

```python
plt.figure()
ax1=plt.axes(projection='3d')
x3=np.linspace(1,50,1000)
y3=np.sin(x3)
z3=np.cos(x3)
plt.plot(x3,y3,z3,c='g')
```
