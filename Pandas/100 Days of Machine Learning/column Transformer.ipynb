```python
from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
```

```python
X_train,X_test,y_train,y_test=train_test_split(covid.drop(columns=['has_covid']),covid['has_covid'],test_size=0.2,random_state=42)
```

```python
#since there are many null values in the fever cols we have to use simple imputer 
si=SimpleImputer(strategy='mean')
X_train_fever=si.fit_transform(X_train[['fever']])
X_test_fever=si.transform(X_test[["fever"]])
```

```python
from sklearn.preprocessing import OrdinalEncoder
from sklearn.preprocessing import OneHotEncoder
```

```python
# ordinal encoding  ==> Cough wala cols 
oe=OrdinalEncoder(categories=[['Mild','Strong']])
#fit and transform to training data 
X_train_cough=oe.fit_transform(X_train[['cough']])
#testing data ma jaile pani transform matra 
X_test_cough=oe.transform(X_train[['cough']])
```

```python
#for nominal data use one hot encoding 
ohe=OneHotEncoder(drop='first',sparse_output=False) # .toarray gari rakhnu parena 
X_train_gender_city=ohe.fit_transform(X_train[['gender','city']])

#for testing data 
X_test_gender_city=ohe.transform(X_test[['gender','city']])
X_train_gender_city.shape
```

```python
from sklearn.compose import ColumnTransformer
```

```python
transformer=ColumnTransformer(transformers=[
  ('tnf1',SimpleImputer(strategy='mean'),['fever']),
  ('tnf2',OrdinalEncoder(categories=[['Mild','Strong']]),['cough']),
  ('tnf3',OneHotEncoder(sparse_output='False',drop='first'),[])
],remainder='passthrough')
```
