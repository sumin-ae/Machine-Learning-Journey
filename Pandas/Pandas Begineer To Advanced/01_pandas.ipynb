### Filtering Of The Data

The most used part of the Data Science and analysis 

```python
#find all the final winners
mask=ipl['MatchNumber']=='Final'  
new_df=ipl[mask]
new_df[['Season','WinningTeam']]                   
```

```python
#One Liner For this 
ipl[ipl["MatchNumber"]=='Final'][['Season','WinningTeam']]
#(filtered_df)[['Season', 'WinningTeam']]
#or more safer way 
ipl.loc[ipl['MatchNumber']=='Final',['Season','WinningTeam']]
#syntax df.loc[row_selector, column_selector]
```

```python
#out of all the match how many super over finishes have occured 
ipl[ipl["SuperOver"]=='Y'].shape[0]
```

```python
ipl.loc[ipl["SuperOver"]=='Y',["Season",'WinningTeam']].shape[0]
```

```python
#how many matches have csk won in kolkata
ipl[(ipl["City"]=='Kolkata') & (ipl["WinningTeam"]=='Chennai Super Kings')].shape[0]
```

```python
ipl.head()
```

```python
#toss winner is the match winner in percentage
#logic behind the implementation 
#ipl["TossWinner"]==ipl['WinningTeam'] gives you a boolean mask then Keeps only the rows where the toss winner also won the match.
wins=ipl[ipl['TossWinner'] == ipl["WinningTeam"]].shape[0]
wins
totalwins=ipl.shape[0]
percent=(wins/totalwins)*100
percent
```
