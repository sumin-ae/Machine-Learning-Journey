<h3> LOC vs ILOC </h3>
LOC- Label Based ==> Give me rows/cols by their name
ILOC- Position Based ==> Give me rows/cols by their position number

```python
coffee.loc[[0,1,2]]
coffee.loc[0:2]
```

```python
coffee.loc[5:10,["Coffee Type","Units Sold"]]
```

```python
coffee.iloc[5:10,[1,2]] 
#notice Smth Different? The Last Row is exclusive in iloc
```
