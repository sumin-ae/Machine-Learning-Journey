<h1> numpy arrays vs python lists </h1>

```python
a = [i for i in range(100000)]
b = [i for i in range(100000, 200000)]
c = []

import time
start_time = time.time()

for i in range(len(a)):
    c.append(a[i] + b[i])

end_time = time.time()
print(f"The total time taken was: {end_time - start_time}")
```

Mathematical Operations in numpy 
<h3> Sigmoid Function </h3> 

```python
def sigmoid(array):
  return 1/(1+np.exp(-(array))) 

random_arr=np.random.randint(1,101,(2,3))
print(random_arr)
sigmoid(random_arr)


```

<h3> Binary Cross Entropy </h3>

```python
def bce(y,yp):
  yp = np.clip(yp, 1e-7, 1 - 1e-7)

  return -(np.mean(y*np.log(yp)+(1-y)*np.log(1-yp)))

actual=np.ones(25) #true value = 1 
predicted=np.arange(25)

bce(actual,predicted)

```

Below are some of the Important Numpy attributes/function for Machine Learning.. 


```python
#np.concatentae
#stiches arrays along a chosen axis 
x= np.array([[1,2],
            [3,4]])
new_features=np.array([[10],[11]])
new_features.shape
x.shape
X_full=np.concatenate([x, new_features],axis=1) #cols are conncatenated
print(X_full)
```

```python
#np.expand_dims => adds a new axis (dimension) of size 1 to a NumPy array at a position you choose.
#adds a new dimension to the array
#My model expects a 2d input but I gave it a 1D- fix its shape 
#this matters because the ml models are picky about shapes 
x= np.array([10,20,40])
x_fixed=(np.expand_dims(x,axis=1)) #do smth at cols 
print(x_fixed)
print(x)
# x.shape       -> (3,)
# x_fixed.shape -> (3, 1)
```

```python
#np.where 
''' If the conficdence of the model is too low, reject prediction
or if the value is missing/bad, replace it'''
confidence=np.array([0.9,0.3,0.6,0,8])
preds=np.array([1,2,3,0,5])
final=np.where(confidence>= 0.5, preds,-1) 
print(final)
''' if the confidence is true than take the corresponding value from the preds and if false replace tyhem with -1'''
```

```python
#np.argmax and np.argmin 
#find the index of the biggest(or smallest) value 
probs = np.array([
    [1, 0.7, 0.2], #row 0
    [0.1, 1, 0.1] #row 1 
])

pred_classes = np.argmax(probs, axis=1)
pred_min_classes=np.argmin(probs,axis=0)
# [0, 1] returns first row ko maximum valued cols 
#returns 2nd rwo ko max value cols element index 
#axis=1 Find the index of the maximum value in each row
#axis=0 FInd the index of the minimum value in each cols 

print(pred_classes)
print(pred_min_classes)

```

```python
#np.clip 
'''  Any values lower than the minimum are set to the minimum, and any values higher than the maximum are set to the maximum.

'''
output=np.arange(-1,11,2)
print(output)
clipped=np.clip(output,0,1)
print(clipped)
```

```python
#np.unique 
#finds all the unique values and optionally counts them 
labels=np.array([1,2,3,4,4,5,5,5,6,6,6,7,7])
classes, counts=np.unique(labels, return_counts=True)
print(f"Classes{classes}\n counts={counts}")
```
